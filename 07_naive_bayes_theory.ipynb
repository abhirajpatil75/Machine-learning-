# Naive Bayes — Theory

## Concept
Naive Bayes applies Bayes’ theorem to estimate class probabilities. Its simplifying assumption is that features are conditionally independent given the class. GaussianNB models continuous features with normal distributions.

**Strengths:** fast, works with small data and multiple classes. **Limitations:** independence assumption may be unrealistic. **Uses:** text categorization and quick baselines.

## Learning task
Classification learns from examples with known categories.

## Main workflow
1. Represent examples as features.
2. Fit model settings on training data (or group unlabeled data).
3. Predict for new examples or inspect discovered groups.
4. Evaluate with metrics suited to the task.

## Metrics
Accuracy is the share correct. Precision summarizes correctness among predicted class members; recall summarizes how many actual class members are found; F1 balances both. Inspect each class.
