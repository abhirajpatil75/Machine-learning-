# Linear Regression — Theory

## Concept
A linear model predicts a numeric target as a weighted sum of input features plus an intercept. It fits weights by minimizing squared prediction errors.

**Form:** prediction = intercept + weighted features.

**Strengths:** simple, fast, interpretable weights. **Limitations:** assumes a linear relationship and can be affected by outliers or correlated features. **Uses:** estimating prices, demand, or other continuous values.

## Learning task
Regression predicts numeric values from examples with known answers.

## Main workflow
1. Represent examples as features.
2. Fit model settings on training data (or group unlabeled data).
3. Predict for new examples or inspect discovered groups.
4. Evaluate with metrics suited to the task.

## Metrics
MAE is average absolute error in target units. RMSE penalizes larger errors more. R² compares predictions with a mean-only baseline; test R² can be negative.
