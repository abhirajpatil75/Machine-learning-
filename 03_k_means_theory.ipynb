# K-Means — Theory

## Concept
K-Means groups examples without known labels. It alternates assigning each point to its nearest centroid and moving each centroid to the mean of its assigned points.

**Objective:** reduce squared distances to centroids. **Strengths:** simple and efficient. **Limitations:** choose k; scaling and outliers matter; it favors round clusters. **Uses:** exploratory grouping and customer segmentation.

## Learning task
Clustering groups observations without supplied target labels.

## Main workflow
1. Represent examples as features.
2. Fit model settings on training data (or group unlabeled data).
3. Predict for new examples or inspect discovered groups.
4. Evaluate with metrics suited to the task.

## Metrics
Inertia measures squared within-cluster distance and always favors larger k. Silhouette score considers cohesion and separation; higher usually means better separated groups.
