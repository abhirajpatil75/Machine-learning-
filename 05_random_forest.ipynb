```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Load the example dataset
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer(as_frame=True)
df = data.frame.copy()
```

```python
# Look at the first rows
df.head()
```

```python
# Column names
df.columns
```

```python
# Dataset size and data types
print("Shape:", df.shape)
df.info()
```

```python
# Summary statistics
df.describe()
```

```python
# Look at the last rows
df.tail()
```

```python
# Check missing values and duplicate rows
print("Missing values by column:\n", df.isna().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Clean the data: remove duplicate rows and fill numeric missing values with column medians
df = df.drop_duplicates().reset_index(drop=True)
df = df.fillna(df.median(numeric_only=True))
print("Clean shape:", df.shape)
print("Missing values remaining:", int(df.isna().sum().sum()))
```

```python
# Analyze numeric columns and plot feature distributions
df.describe().T
df.hist(figsize=(12, 8), bins=20)
plt.tight_layout()
plt.show()
```

```python
# Separate input features (x) and target (y)
x = df.drop(columns="target")
y = df["target"]
```

```python
# Inspect x and y
print(x.head())
print(y.head())
```

```python
# Split into training and testing data
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.20, random_state=42, stratify=y
)
print("Training rows:", len(x_train))
print("Testing rows:", len(x_test))
```

```python
# Scale features for distance- and margin-based models. Fit only on training data.
# Tree and Naive Bayes models use the original feature values in this example.
```

```python
# Create the model
model = RandomForestClassifier(n_estimators=200, random_state=42)
```

```python
# Train the model
model.fit(x_train, y_train)
```

```python
# Make predictions for the test set
y_pred = model.predict(x_test)
```

```python
# Analyze the model results
print("Accuracy:", round(accuracy_score(y_test, y_pred), 3))
print("\nClassification report:\n", classification_report(y_test, y_pred, target_names=data.target_names))
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
# Try a prediction on one held-out example
example = x_test[0:1]
print("Predicted class:", model.predict(example)[0])
print("Actual class:", y_test.iloc[0])
```
