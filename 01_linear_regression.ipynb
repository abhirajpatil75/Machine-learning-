```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Load example numeric regression data
from sklearn.datasets import load_diabetes
data = load_diabetes(as_frame=True)
df = data.frame.copy()
```

```python
# First and last rows
df.head()
```

```python
df.tail()
```

```python
# Missing values and duplicate rows
print(df.isna().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Clean duplicates and fill numeric missing values with medians
df = df.drop_duplicates().reset_index(drop=True)
df = df.fillna(df.median(numeric_only=True))
print("Missing remaining:", int(df.isna().sum().sum()))
```

```python
# Analyze the target and the relationship with one feature
df.describe().T
plt.scatter(df["bmi"], df["target"], alpha=0.6)
plt.xlabel("BMI feature")
plt.ylabel("Disease progression target")
plt.title("Feature and target relationship")
plt.show()
```

```python
# Select x (features) and y (numeric target)
x = df.drop(columns="target")
y = df["target"]
print(x.head())
print(y.head())
```

```python
# Split data before scaling to avoid using test-set information during training
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
print("Training rows:", len(x_train), "Testing rows:", len(x_test))
```

```python
# Scale x using training data only; keep y in its original units
Scaler = StandardScaler()
x_train = Scaler.fit_transform(x_train)
x_test = Scaler.transform(x_test)
```

```python
# Create the Linear Regression model
model = LinearRegression()
```

```python
# Train the model
model.fit(x_train, y_train)
```

```python
# Predict target values for test data
y_pred = model.predict(x_test)
```

```python
# Analyze prediction errors
print("MAE:", round(mean_absolute_error(y_test, y_pred), 2))
print("RMSE:", round(mean_squared_error(y_test, y_pred) ** 0.5, 2))
print("R²:", round(r2_score(y_test, y_pred), 3))
```

```python
# Compare predicted and actual values for one example
print("Predicted:", model.predict(x_test[0:1])[0])
print("Actual:", y_test.iloc[0])
```
