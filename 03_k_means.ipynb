```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
# Load the example dataset
from sklearn.datasets import make_blobs
df = pd.DataFrame(make_blobs(n_samples=400, centers=4, cluster_std=0.75, random_state=42)[0], columns=["feature_1", "feature_2"])
```

```python
# Check missing values and duplicate rows
print("Missing values by column:\n", df.isna().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Clean the data: remove duplicate rows and fill numeric missing values with column medians
df = df.drop_duplicates().reset_index(drop=True)
df = df.fillna(df.median(numeric_only=True))
print("Clean shape:", df.shape)
print("Missing values remaining:", int(df.isna().sum().sum()))
```

```python
# Analyze the clusters
print("Cluster counts:", pd.Series(labels).value_counts().sort_index().to_dict())
print("Inertia:", round(model.inertia_, 2))
print("Silhouette score:", round(silhouette_score(x_scaled, labels), 3))
```
