# Decision Tree — Theory

## Concept
A decision tree asks feature questions in sequence. Each split aims to create purer groups; terminal leaves predict a class.

**Strengths:** visual, handles nonlinear rules, little scaling needed. **Limitations:** deep trees overfit and can change with small data changes. Limit complexity with depth and leaf-size settings. **Uses:** interpretable classification and rule-based grouping.

## Learning task
Classification learns from examples with known categories.

## Main workflow
1. Represent examples as features.
2. Fit model settings on training data (or group unlabeled data).
3. Predict for new examples or inspect discovered groups.
4. Evaluate with metrics suited to the task.

## Metrics
Accuracy is the share correct. Precision summarizes correctness among predicted class members; recall summarizes how many actual class members are found; F1 balances both. Inspect each class.
